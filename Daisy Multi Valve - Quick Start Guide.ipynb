# Daisy Multi Valve - Quick Start Guide

By the end of this notebook you will know how to connect your **Daisy Multi Valve**, home it, and select a port.

## Home the valve

- Always call `valve.home()` before first use. This rotates the valve head to port 1 using the optical end-stop for a reliable reference position.

## Select a port

- Use `valve.select_port(port_number, wait=True)` to rotate to a port with arguments:
>    `port_number` — target port (1–12) \
>    `wait` — if `True` (default), waits for the rotation to complete before running the next line

<br>

**The Daisy Multi Valve uses 1-based indexing, ports are indexed as they are numbered: 1-12.**

## Import

```python
# Packages to install
# Import the daisy python library

import daisylab
import time
```

## Connect and create your instrument group

```python
# Store the Daisy interface in lab
lab = daisylab.connect()

# Store the connected instruments in an instrument group
group = daisylab.InstrumentGroup(lab.instruments_list[0])

# Register your instrument group with the interface
lab.add_group(group)

# Print instruments found and their positions
print("Instruments found:")
for position, (code, inst) in enumerate(zip(lab.instruments_list[0], group.inst)):
    print(f"  [{position}]  {code} — {inst.name}")
```

Change `INST_INDEX` to the position of the Multi Valve you want to use.

```python
# Set which instrument you are using
INST_INDEX = 0   # 0 if first instrument or change to match your multi valve's position

valve = group.inst[INST_INDEX]
print(f"Using: {valve.name}")
```

## Home the valve

Always home before first use. This rotates the valve head to port 1 using the optical end-stop so all port positions are accurate.

```python
valve.home()
print("Homing complete. Valve is at port 1.")
```

```python
print(f"Valve at port: {valve.check_position()}")
```

## Select a port

Change `TARGET_PORT` to the port you want to rotate to (1–12).

```python
# port to rotate to (1–12)
TARGET_PORT = 6

valve.select_port(TARGET_PORT)
print(f"Valve at port: {valve.check_position()}")
```

## Example port selections

```python
# Step through ports 1 to 4 with a pause at each
for port in [1, 2, 3, 4]:
    valve.select_port(port, wait=True)
    print(f"At port {valve.check_position()}")
    time.sleep(1)
```

```python
# Start rotating without waiting
valve.select_port(12, wait=False)
print("Valve is rotating — this line prints immediately")
```

## Close the connection

```python
lab.close()
print("Connection closed.")
```
